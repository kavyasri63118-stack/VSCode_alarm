# 🔔 Laptop Code-Completion Alarm for Jupyter & VS Code
This notebook demonstrates how to get **audio chimes**, **Windows toast notifications**, and **voice alerts** when your notebook cells finish running.

### Step 1: Load the extension

```python
%load_ext code_alarm
```

### Step 2: Use `%%alarm` cell magic on long-running cells
The alarm triggers **only when the cell has completely finished execution** (or if it raises an error).

```python
%%alarm -t "Model Training" -v
import time

print("🧠 Training Machine Learning Model...")
for epoch in range(1, 4):
    print(f"  • Epoch {epoch}/3 complete")
    time.sleep(1)

print("🎉 Model trained successfully!")
```

### Step 3: Enable Global Auto-Alarm (`%alarm_on`)
Automatically chimes & sends a Windows popup for **every cell** in the notebook that runs longer than $N$ seconds:

```python
%alarm_on --min 2
```

```python
# This cell takes 2.5 seconds -> Automatically triggers completion alert!
import time
print("⏳ Processing large dataset...")
time.sleep(2.5)
print("✅ Processing finished!")
```

```python
# To disable global auto-alarm
%alarm_off
```
